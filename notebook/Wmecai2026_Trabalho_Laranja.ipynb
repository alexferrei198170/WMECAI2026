## 1. Introdução

O Brasil é o maior produtor mundial de laranja, sendo o estado de São Paulo responsável pela maior parcela da produção nacional. A comercialização da safra representa uma das decisões mais importantes para o produtor rural.

Este trabalho aplica técnicas de Pesquisa Operacional para apoiar a decisão de destinação da produção entre mercado interno e indústria de suco. Apresenta-se um modelo de Programação Linear para otimizar essa destinação, considerando a sazonalidade dos preços e as janelas operacionais de colheita, respeitando a perecibilidade da fruta.

O modelo foi implementado em Python utilizando GurobiPy e validado por meio de cenários simulados. Os resultados demonstram que o preço pago pela indústria é o fator mais relevante e que a flexibilidade operacional da colheita possui valor econômico mensurável, permitindo ao produtor maximizar sua receita e transformar decisões empíricas em quantitativas.


## 2. Problema de Pesquisa

Um produtor da região de Ribeirão Preto possui produção prevista de 100.000 toneladas de laranja no segundo semestre de 2025. A produção pode ser destinada ao mercado interno, como fruta in natura, ou à indústria de suco. A colheita pode ser postergada em até 10 dias para o mercado interno e até 20 dias para a indústria. O objetivo é maximizar a receita total da propriedade.


## 3. Formulação Matemática

### 3.1 Função Objetivo

A função objetivo busca maximizar a receita total obtida com a destinação da produção ao mercado interno e à indústria de suco:

$$Z = \sum_{t\in T} PM_t M_t + \sum_{t\in T} PI_t I_t$$

em que $M_t$ representa a quantidade destinada ao mercado interno no mês $t$, $I_t$ representa a quantidade destinada à indústria no mês $t$, $PM_t$ é o preço recebido no mercado interno e $PI_t$ é o preço pago pela indústria.

### 3.2 Restrições

**Balanço da produção mensal:**

$$M_t + I_t + TM_t + TI_t = Q_t + TM_{t-1} + TI_{t-1}$$

**Postergação máxima:**

$$TM_t \leq 0{,}20Q_t$$

$$TI_t \leq 0{,}40Q_t$$

**Capacidade do mercado interno:**

$$M_t \leq 12.000$$

**Compromisso mínimo com a indústria:**

$$\sum_{t\in T} I_t \geq 40.000$$

**Não negatividade:**

$$M_t, I_t, TM_t, TI_t \geq 0$$

Nessas expressões, $Q_t$ representa a produção disponível no mês $t$, enquanto $TM_t$ e $TI_t$ representam as quantidades relacionadas à postergação da destinação para o mercado interno e para a indústria, respectivamente.


## 4. Implementação Computacional

O modelo foi implementado em Python utilizando GurobiPy, com apoio das bibliotecas pandas e matplotlib. O método de solução empregado foi o algoritmo Simplex.


## 5. Resultados

As simulações evidenciam dinâmicas importantes. A flexibilidade da colheita tem valor econômico mensurável. Isso mostra que reduzir as janelas de postergação limita a capacidade de aproveitar preços melhores.

A rentabilidade é altamente sensível ao preço industrial. No cenário de aumento de 20% no preço da indústria, a receita sobe para R$ 206,09 milhões, um ganho de R$ 13,85 milhões, ou 7,20%, sobre o cenário base. O volume destinado à indústria aumenta para 76.000 toneladas, evidenciando que esse canal se torna prioritário quando valorizado.

O mercado interno é vulnerável a quedas de preço. O volume destinado ao mercado interno cai para 48.000 toneladas, enquanto a indústria absorve 52.000 toneladas.

O modelo otimizado supera a estratégia tradicional. A estratégia empírica de colher e vender imediatamente gera R$ 197,75 milhões, mas não respeita restrições contratuais e operacionais. Já o modelo otimizado, mesmo com restrições, identifica alocações mais eficientes e cenários de maior ganho.

Os resultados mostram que o preço pago pela indústria é o fator mais relevante. A flexibilidade operacional da colheita possui valor econômico mensurável, e a redução das janelas de colheita gera impacto negativo na receita. A comparação entre o cenário base e o cenário de aumento de 20% no preço industrial evidencia um ganho de rentabilidade de 7,20% para o produtor.


## 6. Conclusão

O conjunto dos cenários evidenciou que a receita citrícola é sensível às variações de mercado e às restrições operacionais. O modelo de Programação Linear proposto, resolvido pelo algoritmo Simplex, mostrou-se capaz de representar o processo decisório, incorporando explicitamente a perecibilidade da fruta e as janelas de postergação no pomar, oferecendo suporte quantitativo à estratégia ótima de comercialização.

```python
from gurobipy import *
import pandas as pd
import matplotlib.pyplot as plt

# FUNÇÃO DE OTIMIZAÇÃO
def executar_modelo(
    nome_cenario,
    preco_mercado,
    preco_industria,
    perc_mercado=0.20,
    perc_industria=0.40
):
    modelo = Model(nome_cenario)
    modelo.setParam("OutputFlag", 0)
    meses = [
        "Julho",
        "Agosto",
        "Setembro",
        "Outubro",
        "Novembro",
        "Dezembro"
    ]
    producao = {
        "Julho": 25000,
        "Agosto": 20000,
        "Setembro": 18000,
        "Outubro": 15000,
        "Novembro": 12000,
        "Dezembro": 10000
    }

    # VARIÁVEIS DE DECISÃO
    M = modelo.addVars(meses, lb=0, name="Mercado")
    I = modelo.addVars(meses, lb=0, name="Industria")
    TM = modelo.addVars(
        meses[:-1],
        lb=0,
        name="TransferenciaMercado"
    )
    TI = modelo.addVars(
        meses[:-1],
        lb=0,
        name="TransferenciaIndustria"
    )

    # FUNÇÃO OBJETIVO
    modelo.setObjective(
        quicksum(
            preco_mercado[m] * M[m]
            for m in meses
        )
        +
        quicksum(
            preco_industria[m] * I[m]
            for m in meses
        ),
        GRB.MAXIMIZE
    )

    # BALANÇO DE PRODUÇÃO
    # Julho
    modelo.addConstr(
        M["Julho"]
        +
        I["Julho"]
        +
        TM["Julho"]
        +
        TI["Julho"]
        ==
        producao["Julho"]
    )
    # Agosto até Novembro
    for i in range(1, 5):
        atual = meses[i]
        anterior = meses[i - 1]
        modelo.addConstr(
            M[atual]
            +
            I[atual]
            +
            TM[atual]
            +
            TI[atual]
            ==
            producao[atual]
            +
            TM[anterior]
            +
            TI[anterior]
        )
    # Dezembro
    modelo.addConstr(
        M["Dezembro"]
        +
        I["Dezembro"]
        ==
        producao["Dezembro"]
        +
        TM["Novembro"]
        +
        TI["Novembro"]
    )

    # POSTERGAÇÃO DA COLHEITA
    for mes in meses[:-1]:
        modelo.addConstr(
            TM[mes]
            <=
            perc_mercado * producao[mes]
        )
        modelo.addConstr(
            TI[mes]
            <=
            perc_industria * producao[mes]
        )

    # CAPACIDADE DO MERCADO INTERNO
    for mes in meses:
        modelo.addConstr(
            M[mes]
            <=
            12000
        )

    # CONTRATO MÍNIMO COM A INDÚSTRIA
    modelo.addConstr(
        quicksum(
            I[m]
            for m in meses
        )
        >=
        40000
    )

    # RESOLUÇÃO
    modelo.optimize()
    if modelo.Status != GRB.OPTIMAL:
        raise Exception(
            f"Não foi encontrada solução ótima para {nome_cenario}"
        )
    receita = modelo.ObjVal
    mercado_total = sum(
        M[m].X
        for m in meses
    )
    industria_total = sum(
        I[m].X
        for m in meses
    )
    return {
        "receita": receita,
        "mercado": mercado_total,
        "industria": industria_total
    }

# RECEITA TRADICIONAL (SEM POSTERGAÇÃO)
receita_tradicional = (
    (12000 * 1700 + 13000 * 1450)
    +
    (12000 * 1850 + 8000 * 1550)
    +
    (12000 * 2000 + 6000 * 1700)
    +
    (12000 * 2150 + 3000 * 1800)
    +
    (12000 * 2250)
    +
    (10000 * 2300)
)

# PREÇOS
mercado_base = {
    "Julho": 1700,
    "Agosto": 1850,
    "Setembro": 2000,
    "Outubro": 2150,
    "Novembro": 2250,
    "Dezembro": 2300
}
industria_base = {
    "Julho": 1450,
    "Agosto": 1550,
    "Setembro": 1700,
    "Outubro": 1800,
    "Novembro": 1850,
    "Dezembro": 1900
}

# CENÁRIO 1 # BASE
cenario1 = executar_modelo(
    "Base",
    mercado_base,
    industria_base
)

# CENÁRIO 2 # MERCADO -15%
mercado_c2 = {
    k: v * 0.85
    for k, v in mercado_base.items()
}
cenario2 = executar_modelo(
    "Mercado -15%",
    mercado_c2,
    industria_base
)

# CENÁRIO 3 # INDÚSTRIA +20%
industria_c3 = {
    k: v * 1.20
    for k, v in industria_base.items()
}
cenario3 = executar_modelo(
    "Indústria +20%",
    mercado_base,
    industria_c3
)

# CENÁRIO 4 # REDUÇÃO DAS JANELAS
cenario4 = executar_modelo(
    "Janela Reduzida",
    mercado_base,
    industria_base,
    perc_mercado=0.10,
    perc_industria=0.20
)

# COMPARATIVO
comparativo = pd.DataFrame({
    "Cenário": [
        "Base",
        "Mercado -15%",
        "Indústria +20%",
        "Janela Reduzida"
    ],
    "Receita": [
        cenario1["receita"],
        cenario2["receita"],
        cenario3["receita"],
        cenario4["receita"]
    ],
    "Mercado Interno (t)": [
        cenario1["mercado"],
        cenario2["mercado"],
        cenario3["mercado"],
        cenario4["mercado"]
    ],
    "Indústria (t)": [
        cenario1["industria"],
        cenario2["industria"],
        cenario3["industria"],
        cenario4["industria"]
    ]
})

# Variação percentual em relação ao cenário Base
comparativo["Variação (%)"] = (
    (comparativo["Receita"] - cenario1["receita"]) / cenario1["receita"]
) * 100

# Criar versão formatada para exibição
comparativo_fmt = comparativo.copy()
comparativo_fmt["Receita (R$ milhões)"] = comparativo_fmt["Receita"].map(lambda x: f"R$ {x/1e6:,.2f}")
comparativo_fmt["Variação (%)"] = comparativo_fmt["Variação (%)"].map(lambda x: f"{x:.2f}%")

# Remover coluna Receita bruta (se quiser apenas a formatada)
comparativo_fmt = comparativo_fmt.drop(columns=["Receita"])

print("\n")
print("=" * 100)
print("COMPARATIVO DOS CENÁRIOS")
print("=" * 100)
display(comparativo_fmt)

# GRÁFICO
plt.figure(figsize=(10,6))

# Cores específicas para cada cenário
cores = ["orange","gold","green","blue"]

# Barras
plt.bar(comparativo["Cenário"], comparativo["Receita"]/1e6, color=cores)

# Título e eixos
plt.title("Receita Máxima por Cenário")
plt.ylabel("Receita (Milhões de R$)")
plt.grid(axis="y", alpha=0.3)

# Valores acima das barras
for i, valor in enumerate(comparativo["Receita"]):
    plt.text(i, valor/1e6 + 0.5, f"{valor/1e6:.2f}", ha="center")



plt.tight_layout()
plt.show()

# CONCLUSÕES AUTOMÁTICAS
melhor = comparativo.loc[
    comparativo["Receita"].idxmax()
]
print("\n")
print("=" * 100)
print("CONCLUSÕES")
print("=" * 100)
print(
    f"Melhor Cenário: "
    f"{melhor['Cenário']}"
)
print(
    f"Receita Máxima: "
    f"R$ {melhor['Receita']:,.2f}"
)
print("\nINTERPRETAÇÃO")
print(
    "- O modelo identifica a alocação ótima entre mercado interno e indústria."
)
print(
    "- A postergação limitada da colheita possui valor econômico."
)
print(
    "- A indústria torna-se mais atrativa quando seus preços aumentam."
)
print(
    "- A redução das janelas operacionais diminui a flexibilidade do sistema."
)
print(
    "- A Programação Linear mostrou-se eficaz como ferramenta de apoio à decisão."
)
```
